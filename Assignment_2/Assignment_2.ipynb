# Background

## DNA

### Sugar-phosphate backbone
* Two chains of sugars connected via phosphates (sugar-phosphate backbone) form the structural framework of DNA
* The two ends of each backbone chain are distinct and lend their names to defining DNA strand orientation (see below): the 5' ("five prime") end has a free phosphorus group and the 3' end has a free hydroxyl (OH) group

### Complementarity
* Along each sugar-phosphate chain, four bases (Adenine, Thymine, Guanine, and Cytosine) make up the information stored in each molecule
* The two strands are held together by hydrogen bonds between these bases, with the backbones to the outside, forming a double helix
* The bases from each strand form complementary base pairs: 
    - Adenine (A) pairs with Thymine (T), using 2 hydrogen bonds
    - Guanine (G) pairs with Cytosine (C), using 3 hydrogen bonds
* Therefore, each strand forms a kind of inverted mirror image of the other - you can deduce the complementary strand by base pairing

### Strand orientation
* The two DNA strands forming a double helix run antiparallel to each other (i.e., the 5' end of one strand binds to the 3' end of the other)

### Packaging of DNA into chromosomes

- The DNA double helix winds around histone proteins to form nucleosomes ("beads on a string")
- This string gets tightly packed into chromatin fiber, which then forms chromosomes

![](https://media.geeksforgeeks.org/wp-content/uploads/20230504103842/DNA-Packaging.jpg)

## RNA
- Beside DNA, there is another nucleic acid present in cells: RNA, which can express (transcribe) the genetic instructions activated in each cell. It is involved as an intermediate step in protein synthesis. 

- RNA is a single-stranded molecule produced from a DNA template by base matching, substituting Uracil (U) for Thymine (T). 

- The three main RNA types involved in protein synthesis are: 
  - messenger RNA (mRNA): acts as a transcript copy of a gene, forms the template for the ribosome
  - transfer RNA (tRNA): carries amino acids to the ribosome during the translation process
  - ribosomal RNA (rRNA): forms the ribosome, aligns mRNA and tRNA to correctly match three-base nucleotide sequences (codons) with their corresponding amino acids

## Proteins
- Proteins are composed of chains of amino acids (polypeptides), assembled by the ribosome following the sequence specified in an mRNA molecule
- There are 20 main amino acids which form the basic alphabet of proteins, universal to all organisms
- They all contain the same central structure and differ in a variable side chain (residue), which determines the chemical (functional) properties of the molecule

## The Genetic Code
- The sequence of nucleotides in an mRNA molecule is translated into a sequence of amino acids using a code: a correspondence of nucleotide triplets ("codons") to amino acids
- There are 64 $=4^3$ possible codons, as shown in the following "codon sun" (read from the inside to the outside, i.e. 5' to 3')
- The start of a "coding" nucleotide sequence in a long string of RNA is always given by `AUG` (the start codon), and the end is given by one of three end codons (`UAA`, `UAG`, or `UGA`)
    - This coding sequence is called an Open Reading Frame (ORF)
    - There are six possible reading frames from a double-stranded DNA molecule: three on each single strand
- Clearly, there is some degeneracy in this code: multiple codons encode the same amino acid
    - Degeneracy is restricted to the third nucleotide of the codon, as it is less strictly paired to the tRNA providing the corresponding amino acid
    - This provides some protection against mutation effects ("silent mutations" in the third codon base do not affect the resulting protein)
